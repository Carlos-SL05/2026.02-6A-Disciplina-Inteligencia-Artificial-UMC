Utilizei um loop para verificar cada lista da matriz e dentro desse loop, um outro loop para verificar cada valor dentro da lista, com isso cada iteração realiza a multiplicação de cada número da matriz pela sua posição.

```python
import numpy as np # Importa a biblioteca numpy
```

```python
matriz = np.array(np.random.randint(1, 10, (4, 4))) # Criando a matriz

multiplicada = matriz.astype(int) # Cria uma matriz independente com os mesmos valores da primeira matriz

for linha in range(4):
  for coluna in range(4):
    multiplicada[linha, coluna] = matriz[linha, coluna] * linha * coluna # Multiplica o número da matriz pela sua posição


print(f'Matriz: \n{matriz}\n') # Imprime a matriz

print(f'Matriz multiplicação: \n{multiplicada}') # imprime a matriz multiplicada
```

#Exercício NumPy 19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).
---
#Justificativa

Primeiro importei a biblioteca numpy apelidando de np, utilizando np.array para criar um array e np.random.randint para gerar os números inteiros aleatórios, depois utilizei np.dot() para calcular o produto escalar e depois o np.sqrt() calcula as raízes quadradas de cada array, definindo assim as normas para o resto do cálculo, com isso o valor do cosseno é definido pela divisão do produto escalar pela multiplicação das normas. No final utilizo np.arccos() para ter o valor em radianos e np.degrees para ter o valor em graus.

```python
import numpy as np # Importa a biblioteca numpy
```

```python
# Defini os dois arrays
array1 = np.array(np.random.randint(1, 21, 20))
array2 = np.array(np.random.randint(1, 21, 20))

produto_escalar = np.dot(array1, array2) # Calcula o valor escalar

# Defini as normas
norma1 = np.sqrt(np.sum(array1 ** 2))
norma2 = np.sqrt(np.sum(array2 ** 2))

cos = produto_escalar / (norma1 * norma2) # Calcula o cosseno

angulo_rad = np.arccos(cos) # Defini o valor do ângulo em radianos
angulo_graus = np.degrees(angulo_rad) # Defini o valor do ângulo em graus

print(f'Array1: {array1}\n')
print(f'Array2: {array2}\n')
print(f'norma1: {norma1}\n')
print(f'norma2: {norma2}\n')
print(f'cosseno: {cos}\n')
print(f'radiano: {angulo_rad}\n')
print(f'graus: {angulo_graus}')
```

#Exercício SciPy 4. Use a função de otimização de SciPy para maximizar f(x) = −x²+4x+1
---
#Justificativa
---
Primeiro importei minimize da biblioteca scipy.
Defini a função de x, retornando -x² + 4 vezes x + 1, depois foi definido a função negativa, com isso foi utilizado minimize() para minimizar a função negativa e por fim, o valor máximo é o oposto ao valor mínimo encontrado, nesse caso armazenado em resultado.fun, portanto -resultado.fun retorna o valor máximo.

```python
from scipy.optimize import minimize # Importa minimize do módulo optimize da biblioteca scipy
```

```python
# Define a função
def f(x):
  return -x**2 + 4*x + 1

# Define a função negativa da função original
def funcao_negativa(x):
  return -f(x)

resultado = minimize(funcao_negativa, 0) # Minimiza a função negativa

valor_maximo = -resultado.fun # Define o valor máximo, que é o oposto do valor mínimo encontrado

# Imprime os resultados
print(f'Resultado: {resultado.x[0]:.2f}')
print(f'Valor máximo: {valor_maximo:.2f}')
```

Defini o sistema de equação, o intervalo de x e os valores de y(0) e y'(0), após isso, foi utilizado o solve_ivp() para a resolução da equação, passando os valores da equação, do intervalo de x, os valores de y e 100 pontos para o intervalo de 0 a 10.

```python
from scipy.integrate import solve_ivp # Importa solve_ivp da biblioteca scipy
```

```python
import numpy as np # Importa a biblioteca numpy
```

```python
# Define o sistema de equações
def eq_diferencial(x, y):
  y1, y2 = y
  dy1_dx = y2
  dy2_dx = (-y1)
  return [dy1_dx, dy2_dx]

x_span = (0, 10) # Define intervalo de x de 0 a 10
y0 = [1, 0] # Define valores de y(0) e y'(0)

solucao = solve_ivp(eq_diferencial, x_span, y0, t_eval=np.linspace(0, 10, 100)) # Solução da equação

# Imprime os resultados
print(f'Valores de X: \n{solucao.t}\n')
print(f'Valores de Y(0): \n{solucao.y[0]}\n')
print(f"Valores de y'(0): \n{solucao.y[1]}")
```

#Justificativa

Primeiro importei a biblioteca pandas, apelidando de pd.
Defini os dados, depois criei dois DataFrames baseados nesses dados com pd.DataFrame(), depois usei pd.merge() para mesclar os dois DataFrames a partir da coluna id, juntando as duas tabelas onde os ids são iguais.

```python
import pandas as pd # Importa a biblioteca pandas
```

```python
dados1 = {
    'id': [0, 1, 2, 3, 4],
    'Produto': ['Huntr/x photocard', 'Hollow Knight: Silksong', 'Hollow Knight', 'Ado plushie', 'Ado lightstick'],
    'Preço': [8.00, 59.99, 46.99, 124.30, 170.00],
    'Quantidade': [23, 15, 13, 10, 8],
    'Categoria': ['Photocards', 'Jogos', 'Jogos', 'Pelúcias', 'Lightsticks']
} # Fornecendo dados

dados2 = {
    'id': [0, 1, 2, 3, 4],
    'descricao': ['Pacote com 3 photocards do grupo huntr/x', 'Jogo Hollow Knight: Silksong', 'Jogo Hollow Knight', 'Pelúcia da cantora Ado', 'Lightstick da cantora Ado']
} # Fornecendo os dados

df1 = pd.DataFrame(dados1) # Criando o primeiro DataFrame

df2 = pd.DataFrame(dados2) # Criando o segundo DataFrame

mescla = pd.merge(df1, df2, on='id') # Mescla os dois DataFrames

print(mescla) # Mostra o DataFrame
```

#Exercício pandas 20. Agrupe os dados por uma coluna categórica e calcule a mediana de uma coluna numérica
---
#Justificativa

Primeiro importei a biblioteca pandas, apelidando de pd.
Defini os dados, depois criei um DataFrame baseado nesses dados com pd.DataFrame(), depois usei df.groupby('Categoria') para agrupar os dados de Categoria e por fim usei esses valores agrupados e usei ['Preço'].median() para que o valor da mediana de Preço fosse encontrado

```python
import pandas as pd # Importa a biblioteca pandas
```

```python
dados = {
    'Produto': ['Huntr/x photocard', 'Hollow Knight: Silksong', 'Hollow Knight', 'Ado plushie', 'Ado lightstick'],
    'Preço': [8.00, 59.99, 46.99, 124.30, 170.00],
    'Quantidade': [23, 15, 13, 10, 8],
    'Categoria': ['Photocards', 'Jogos', 'Jogos', 'Pelúcias', 'Lightsticks']
} # Fornecendo dados

df = pd.DataFrame(dados) # Criando o DataFrame

df_agrupado = df.groupby('Categoria')['Preço'].median()  # Agrupa o DataFrame pela coluna Categoria e calcula a mediana da coluna preço

print(df_agrupado) # Mostra o DataFrame agrupado pela categoria e com a mediana de preço
```

#Exercício Matplotlib 8. Plote um gráfico de linha com intervalos de confiança (sombramento) para a função y = sin(x).
---
#Justificativa

Primeiro importei pyplot de matplotlib e apelidei de plt e também importei numpy, apelidando de np.
Defini x como valores de 0 a 10 e y como seno de x, após isso defini y_inferior como 0,2 a menos que y e y_superior como 0,2 a mais que y, depois foi criado um gráfico de linha com os valores de x e y usando plt.plot(x, y) e por fim para aplicar os intervalos de confiança foi utilizado o plt.fill_between(x, y_inferior, y_superior)

```python
import matplotlib.pyplot as plt # Importa pyplot da biblioteca matplotlib e apelida de plt
```

```python
import numpy as np # Importa a biblioteca numpy e apelida de np
```

```python
# Definindo os eixos x e y
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Define os intervalos
y_superior = y + 0.2
y_inferior = y - 0.2

plt.plot(x, y) # Cria o gráfico de linha

plt.fill_between(x, y_inferior, y_superior) # Preenche a área entre os intervalos

plt.title('Gráfico de sombramento') # Cria um título para o gráfico
plt.xlabel('Eixo x') # Nomeia o eixo x
plt.ylabel('Eixo y') # Nomeia o eixo y

plt.show() # Mostra o gráfico
```

#Exercício Matplotlib 13. Gere um gráfico de dispersão 3D para 100 pontos aleatórios utilizando Matplotlib.
---
#Justificativa

Primeiro importei pyplot de matplotlib e apelidei de plt e também importei numpy, apelidando de np.
Defini x, y e z como 100 números aleatórios entre 0 e 1, depois foi criado uma janela de figura em branco com plt.figure(), após isso foi utilizado .add_subplot() para adicionar um subplot na figura, definindo a projeção como 3D com projection='3d' e o 1,1,1 define uma grade 1x1 e que será colocado no primeiro gráfico(nesse caso o único), definido a projeção 3D, o gráfico de dispersão é criado com plt.scatter() com o marker sendo 'o' para que os pontos sejam em formato de círculos.

```python
import matplotlib.pyplot as plt # Importa pyplot da biblioteca matplotlib e apelida de plt
```

```python
import numpy as np # Importa a biblioteca numpy e apelida de np
```

```python
# Definindo os eixos x, y e z
x = np.array(np.random.rand(100))
y = np.array(np.random.rand(100))
z = np.array(np.random.rand(100))

fig = plt.figure() # Gera uma figura

grafico_3d = fig.add_subplot(1, 1, 1, projection='3d') # Adiciona um subplot na figura com projeção 3D

grafico_3d.scatter(x, y, z, marker='o') # Cria o gráfico de dispersão

plt.title('Gráfico de dispersão 3D') # Cria um título para o gráfico
plt.xlabel('Eixo x') # Nomeia o eixo x
plt.ylabel('Eixo y') # Nomeia o eixo y

plt.show() # Mostra o gráfico 3D
```

#Exercício Matplotlib 18. Crie um gráfico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor.
---
#Justificativa

Primeiro importei pyplot de matplotlib e apelidei de plt.
Defini as categorias e os valores, após isso, as cores são definidas por meio de uma condicional onde cada valor dentro de valores é verificado e caso o valor seja menor que zero, a cor recebida é vermelha, caso contrário, a cor recebida será azul, por fim, o gráfico de barras é criado com as categorias, valores e cores e no final o plt.show() mostra o gráfico gerado.

```python
import matplotlib.pyplot as plt # Importa pyplot da biblioteca matplotlib e apelida de plt
```

```python
# Definindo os eixos x e y
categorias = ['A', 'B', 'C', 'D']
valores = [-10, -5, 5, 10]

cores = ['red' if valor < 0 else 'blue' for valor in valores] # Definindo as cores

plt.bar(categorias, valores, color=cores) # Criando o gráfico de barras com as cores aplicadas

plt.show() # Mostra o gráfico
```

#Exercício Matplotlib 19. Plote um gráfico de linha com duas séries e adicione uma legenda para diferenciá-las.
---
#Justificativa

Primeiro importei pyplot de matplotlib e apelidei de plt e também importei numpy, apelidando de np.
Defini x com valores de 0 a 10, y1 com seno de x e y2 com cosseno de x, depois utilizei dois plt.plot() para criar gráficos de linhas, um com os valores de x e y1 e outro com os valores de x e y2, utilizando label= em cada um para dar um rótulo, nomeando de Seno o primeiro e Cosseno o segundo.

```python
import numpy as np # Importa a biblioteca numpy e apelida de np
```

```python
# Definindo os eixos x e y
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Cria os gráficos de linha
plt.plot(x, y1, label='Seno')
plt.plot(x, y2, label='cosseno')

plt.title('Gráfico com duas séries') # Cria um título para o gráfico
plt.xlabel('Eixo x') # Nomeia o eixo x
plt.ylabel('Eixo y') # Nomeia o eixo y

plt.legend() # Cria a legenda do gráfico

plt.show() # Mostra o gráfico
```

Defini x com valores de 0 a 10 e y com x ao quadrado, depois usei plt.plot() para criar um gráfico de linha com os valores de x e y. **PREENCHER**. Utilizei plt.xticks() para definir o espaçamento dos números do eixo x, indo de -10 a 10 e indo de 2 em 2 e o plt.yticks() para definir o espaçamento dos números do eixo y, indo de 0 a 100 pulando de 20 em 20 números, por fim o plt.grid() foi usado para mostrar a grade no gráfico e com a cor roxa

**PREENCHER**

```python
import matplotlib.pyplot as plt # Importa pyplot da biblioteca matplotlib e apelida de plt
```

```python
import numpy as np # Importa a biblioteca numpy e apelida de np
```

```python
# Definindo os eixos x e y
x = np.linspace(0, 10, 100)
y = x ** 2

plt.plot(x, y) # Cria o gráfico de linha

plt.title('Gráfico com espaçamento de grade personalizado') # Cria um título para o gráfico
plt.xlabel('Eixo x') # Nomeia o eixo x
plt.ylabel('Eixo y') # Nomeia o eixo y

plt.xticks(np.arange(-10, 11, 2)) # Delimita o espaçamento do eixo x de -10 a 10, pulando de 2 em 2
plt.yticks(np.arange(0, 101, 20)) # Delimita o espaçamento do eixo y de 0 a 100, pulando de 20 em 20

plt.grid(
    visible=True,
    color='purple'
) # APlica a grade visível e com a cor roxa

plt.show() # Mostra o gráfico
```
